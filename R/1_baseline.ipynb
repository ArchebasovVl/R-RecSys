```r
source("./scripts/utils.r")

df <- read.csv("./data/fashion_products.csv")

df[, "User.ID"] <- as.character(df[, "User.ID"])
df[, "Product.ID"] <- as.character(df[, "Product.ID"])

sapply(df, class)
cat("\n")
head(df)
```

```r
result <- create_user_item_matrix(df, "User.ID", "Product.ID", "Rating")
user_rating_matrix <- result[1][[1]]
user_map <- result[2][[1]]
item_map <- result[3][[1]]

print(dim(user_rating_matrix))
```

```r
model <- fit_baseline(user_rating_matrix)
pred <- predict_baseline(model, user_rating_matrix, '1', user_map)
best_ind <- choose_k_best(pred, 3)

print("best products ids for the first user:")
for (i in best_ind) {
    cat("id:", item_map[[as.character(i)]], "\tevaluated rating:", pred[i], "\n")
}
```
